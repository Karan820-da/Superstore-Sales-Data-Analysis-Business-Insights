```python
# Converting Order dates and Ship dates into date fromat 

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
```

```python
df.groupby('region')['sales'].sum()
```

```python
df.groupby('category')['profit'].sum().sort_values(ascending=False)
```

```python
df.groupby('sub-category')['sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('customer_id')['sales'].sum().sort_values(ascending=False).head(3).tail(2)
```

```python
top_customer = (
    df.groupby(['region','customer_name'])['sales']
    .sum()
    .reset_index()
    .sort_values(['region','sales'], ascending=[True, False])
    .groupby('region')
    .head(1)
)
```
